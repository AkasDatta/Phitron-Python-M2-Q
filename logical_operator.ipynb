Logical Operator

```python
x = 10; y = 209; z = 3 # and এর ক্ষেত্রে দুইটি শর্তই সত্য হতে হবে। যদি একটিও শর্ত মিথ্যা হয় তাহলে ফলাফল মিথ্যা হবে।

result = (x > y) and (x < z)
print (result)
```
